```python
import numpy as np
import pandas as pd
import random
import matplotlib.pyplot as plt

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Dense, Flatten
```

```python
X_train = pd.read_csv('input.csv', header=None, on_bad_lines='skip').values
Y_train = pd.read_csv('labels.csv', header=None).values

X_test = pd.read_csv('input_test.csv', header=None, on_bad_lines='skip').values
Y_test = pd.read_csv('labels_test.csv', header=None).values

# Align sizes (VERY IMPORTANT)
min_len_train = min(len(X_train), len(Y_train))
X_train, Y_train = X_train[:min_len_train], Y_train[:min_len_train]

min_len_test = min(len(X_test), len(Y_test))
X_test, Y_test = X_test[:min_len_test], Y_test[:min_len_test]
```

```python
assert X_train.shape[1] == 30000
```

```python
X_train = X_train.reshape(-1, 100, 100, 3).astype('float32') / 255.0
X_test = X_test.reshape(-1, 100, 100, 3).astype('float32') / 255.0

Y_train = Y_train.reshape(-1, 1)
Y_test = Y_test.reshape(-1, 1)
```

```python
print("X_train:", X_train.shape)
print("Y_train:", Y_train.shape)
print("X_test :", X_test.shape)
print("Y_test :", Y_test.shape)
```

```python
idx = random.randint(0, len(X_train) - 1)

plt.imshow(X_train[idx])
plt.title(f"Label: {Y_train[idx][0]}")
plt.axis('off')
plt.show()
```

```python
model = Sequential([
    Conv2D(32, (3,3), activation='relu', input_shape=(100, 100, 3)),
    MaxPooling2D(2,2),

    Conv2D(32, (3,3), activation='relu'),
    MaxPooling2D(2,2),

    Flatten(),
    Dense(64, activation='relu'),
    Dense(1, activation='sigmoid')
])
```

```python
model.compile(
    loss='binary_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)
```

```python
model.fit(X_train, Y_train, epochs=5, batch_size=64)
```

```python
model.evaluate(X_test, Y_test)
```

```python
idx2 = random.randint(0, len(X_test) - 1)

plt.imshow(X_test[idx2])
plt.axis('off')
plt.show()

y_pred = model.predict(X_test[idx2].reshape(1, 100, 100, 3))
y_pred = (y_pred > 0.5).astype(int)

if y_pred[0][0] == 0:
    print("Prediction: dog")
else:
    print("Prediction: cat")
```
